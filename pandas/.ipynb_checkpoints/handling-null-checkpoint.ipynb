```python
df=pd.read_csv("weather_data.csv",parse_dates=['day'])#coverts value in day col to date type
type(df['day'][0])
```

```python
#convert date in index
df.set_index("day",inplace=True)
df
```

```python
df.fillna(0,inplace=True) 
df
```

```python
df
```

```python
df.fillna({
    'temperature':df.temperature.mean(),
    'windspeed' : df.windspeed.mean()
})
```

```python
df = df.fillna({
    'temperature': df['temperature'].mean(),
    'windspeed': df['windspeed'].mean(),
    'event' : "No event"
})

df
```

```python
df.fillna(method='ffill')#forward fill fill the prv row value of the col in null value
```

```python
df.fillna(method='bfill')
```

```python
df.fillna(method='bfill',axis='columns') # it will fill the null values with value in the next col of the same row
```

```python
df.fillna(method='ffill',axis='columns') # it will fill the null values with value in the prev col of the same row
```

```python
df.fillna(method='ffill',limit=1)#suppose therre are contious null values and u want to fil 1 nul value wiht the prev val
```

```python
df.interpolate() #gives middle val of next and prev
```

```python
df.dropna()
```

```python
df.dropna(how='all') #if all val are na
```

```python
df.dropna(thresh=2)#drop if atleast 2 na values
```
