# Read Excel

```python
#excel file has multiple sheets
df_movies=pd.read_excel("movies_db.xlsx","movies") #movies is the sheet name
df_movies
```

```python
df_financials=pd.read_excel("movies_db.xlsx","financials") #financials is the sheet name
df_financials.head()
```

```python
def standardize_currency(curr):
    if curr == '$$' or curr == 'Dollars':
        return "USD"
    return curr

df_financials = pd.read_excel("movies_db.xlsx", "financials", converters = {
    'currency': standardize_currency
})
df_financials.head(5)
```

```python
df_merged=pd.merge(df_movies,df_financials,on="movie_id")
df_merged
```

```python
#exporting excel fFile 
df_merged.to_excel("movies_merged.xlsx",sheet_name="Sheet1",index=False)
```

```python
df_stocks = pd.DataFrame({
    'tickers': ['GOOGL', 'WMT', 'MSFT'],
    'price': [845, 65, 64 ],
    'pe': [30.37, 14.26, 30.97],
    'eps': [27.82, 4.61, 2.12]
})

df_weather =  pd.DataFrame({
    'day': ['1/1/2017','1/2/2017','1/3/2017'],
    'temperature': [32,35,28],
    'event': ['Rain', 'Sunny', 'Snow']
})
```
