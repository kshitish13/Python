# Decorator

A decorator is a function that adds extra functionality to another function without changing the original function's code.

Original function \
          ↓  \
   Decorator \
          ↓  \
Function with extra functionality

```python
import time

def get_square(numbers):
    start = time.time()
    result = []
    for n in numbers:
        result.append(n*n)
    end = time.time()
    print(f"get_square took {(end - start)*1000} mil seconds")
    return result
```

```python
def get_cube(numbers):
    start = time.time()
    result = []
    for n in numbers:
        result.append(n*n*n)
    end = time.time()    
    print(f"get_cube took {(end - start)*1000} mil seconds")
    return result
```

#### Here we are repeating the code of start timr and end time and printing 
#### Suppose we have 100s of function repeatation like this would be inefficient


```python
mylist = [1,2,3,4,5]

get_square(mylist)
```

```python
get_cube(mylist)
```

#### Let's try a big python list

```python
_ = get_square(range(100000))
```

```python
_ = get_cube(range(100000))

```

#### Here we are repeating the code of start timr and end time and printing 
#### Suppose we have 100s of function repeatation like this would be inefficient
#### SO python provides a way to optimize this and that is decorator

So we can use some annotation like @timer which is just another function and the timer function is a decorator

```python
def timer(func): 
    def wrapper(*args, **kwargs): 
        start=time.time()
        result = func(*args,**kwargs)
        end=time.time()
        print(f"Function {func.__name__} took {(end-start)*1000} seconds")
        return result
    return wrapper
```

```python
@timer
def get_square(numbers):
    result = []
    for n in numbers:
        result.append(n*n)
    return result
```

```python
_ = get_square(range(10000))
```
