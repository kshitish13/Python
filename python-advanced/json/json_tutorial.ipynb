#### Converting from Json String to Dict

```python
# JSON string
json_string = '''
{
    "name": "Maulik",
    "age": 30,
    "is_employee": true,
    "skills": ["Python", "Data Analysis", "Machine Learning"],
    "address": {
        "street": "123 Main St",
        "city": "Wonderland"
    }
}
'''
```

```python
#c0verting json to dictionary
import json

record = json.loads(json_string)
record
```

```python
type(record)
```

```python
record['name']
```

#### Converting from Dict to JSON string

```python
data = {
    "name": "Bob",
    "age": 25,
    "is_employee": False,
    "skills": ["JavaScript", "Web Development"],
    "address": {
        "street": "456 Elm St",
        "city": "Springfield"
    }
}

```

```python
json_str=json.dumps(data)
json_str
```

```python
type(json_str)
```

#### Write JSON data  to file

```python
with open("output.json",'w') as f:
    json.dump(data,f,indent=4)
```

#### Read JSON data from file

#### Read JSON data from file

```python
json_string = '''
{
    "books": [
        {"title": "The Magic Of Thinking Big", "author": "David J. Schwartz", "published": 1959},
        {"title": "Chanakya Neeti", "author": "B.K. Chaturvedi", "published": 2009},
        {"title": "How to Win Friends and Influence People", "author": "Dale Karnegie", "published": 1936}
    ],
    "status_code": 200
}
'''
```

```python
data=json.loads(json_string)
data
```

```python
books=data['books']
books
```

```python
books_list=[[book['title'], book['author']] for book in books]
books_list
```

```python
books_list = []

for book in books:
    books_list.append({book['title'] : book['author']})

print(books_list)
```

```python
books_list=[{'title' : book['title'],'author' : book['author']} for book in books]
books_list
```

#### Error Handling in JSON

```python
# Malformed JSON string
json_string = '''
{
    "name": "Krishna",
    "age": 30
    "is_employee": true
}
'''
```

```python
json.loads(json_string)
```

```python
try:
    data = json.loads(json_string)
except json.JSONDecodeError as e:
    print(f"Failed to parse JSON: {e}")
```
