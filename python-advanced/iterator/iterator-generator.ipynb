```python
#we will creatw a custom remote control class when do next it should give the next chqnnel

class Remote: 
    def __init__(self): 
        self.channels=['NDTV','CNN','HBO']
        self.index=-1

    def __next__(self): #It tells Python:
                    #"When someone asks for the next item, this is what you should do."
        self.index+=1
        if self.index==len(self.channels): 
            raise StopIteration
        return self.channels[self.index]

    def __iter__(self):
        return self
```

```python
rc=Remote()
next(rc)
```

# Generator

A generator is a special type of function that produces values one at a time, instead of creating and storing all the values at once.

```python
def remote_control(): 
    yield "NDTV"
    yield "CNN"
    yield "ESPN"
```

```python
rc=remote_control()
type(rc)
```

```python
for channel in rc: 
    print(channel)
```

```python
#We can get values using next():
next(rc)
```

```python

def get_company_data(): 
    data=[]
    for i in range(1,4):
        with open(f"stock_{i}.txt","r") as f:
            content=f.read()
            data.append(content)
    return data
```

```python
get_company_data() 
```

* <h4>when the above code was executed in the company it was going out of memory because the data was too large </h4>
* <h4>so we need to  use generator so that it does not give all data at a time and gives it one at a time </h4>

```python
def get_company_data(): 
    data=[]
    for i in range(1,4):
        with open(f"stock_{i}.txt","r") as f:
            content=f.read()
            yield content
```

```python
for company_data in get_company_data(): 
    print(company_data)
```
