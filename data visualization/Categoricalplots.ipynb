# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this: 

* boxplot
* violinplot
* stripplot
* stripplot
* barplot
* countplot

Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

```python
df.iloc[67]
```

```python
df['size'].unique()
```

```python
sns.countplot(x = df['sex'])
```

```python
sns.countplot(x = df['sex'],hue = df['smoker'])
```

```python
sns.countplot(x = df['smoker'],hue = df['sex'])
```

```python
sns.barplot(x = df['sex'],y = df['total_bill'])
```

You can change the estimator object to your own function, that converts a vector to a scalar Mean (default): np.mean Median: np.median Sum: np.sum Count: len (to count the number of observations) Standard Deviation: np.std Minimum: np.min Maximum: np.max

```python
import numpy as np 
sns.barplot(x = df['sex'],y = df['tip'],estimator=np.sum) #by default estimator = mean
```
