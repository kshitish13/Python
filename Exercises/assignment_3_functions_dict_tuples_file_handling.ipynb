### Task 2

```

```

````python
# write your code here
am=float(input("Enter the purchase amount :"))

def calculate_rewards(total):
    if total >= 500:
        return 'Gold'
    elif total >= 200:
        return 'Silver'
    elif total >= 100:
        return 'Bronze'
    else:
        return 'None'
calculate_rewards(am)
````

````python
# write your code here
def give_summary(file):
    customer_detail={}
    with open(file,"r") as f:
        for line in f:
            name,amount=line.split(",")
            amount=int(amount.strip())
            customer_detail[name]=amount

    def calculate_rewards(total):
        if total >= 500:
            return 'Gold'
        elif total >= 200:
            return 'Silver'
        elif total >= 100:
            return 'Bronze'
        else:
            return 'None'

    for key,value in customer_detail.items():
        customer_prize=calculate_rewards(value)
        customer_detail[key]=(value,customer_prize)

    return customer_detail
````
