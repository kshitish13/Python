Example URL for Texas Breweries: `https://api.openbrewerydb.org/v1/breweries?by_state=texas`

````python
# write your code here
import requests

url='https://api.openbrewerydb.org/v1/breweries?by_state=texas'
response=requests.get(url)

if response.status_code==200 :
    data=response.json()
    print(data)
else:
    print("Failed to fetch data")
````

````python
def fetch_breweries_by_state(state): 
    url=f'https://api.openbrewerydb.org/v1/breweries?by_state={state}'

    try:
        response=requests.get(url)
        
             # Raise an error if the API returns 4xx or 5xx
        response.raise_for_status() #it is necessary as requests.get() does not automatically raise an exception for HTTP errors such as 404 or 500.
        
        breweries=response.json()

        if breweries:
            for data in breweries: 
                name=data['name']
                brewery_type=data['brewery_type']
                city=data['city']

                print(f'{name} of {brewery_type} is located in {city}')

        else:
            print("No breweries found")

    except requests.exceptions.HTTPError as err:
        return f"HTTP error occurred: {err}"
    except Exception as err:
        return f"An error occurred: {err}"

fetch_breweries_by_state('New York')
````

* <h3>Usually, a function named fetch should fetch/get the data and return it.</h3>
* <h3>printing the data in function doesn't return the brewery data.</h3>
* <h3>So return the brewery in function and print outside</h3>
<h3></h3>

````python
import requests

def fetch_breweries_by_state(state):
    url = f"https://api.openbrewerydb.org/v1/breweries?by_state={state}"

    try:
        response = requests.get(url)
        response.raise_for_status()

        breweries = response.json()

        return breweries

    except requests.exceptions.HTTPError as err:
        print(f"HTTP error occurred: {err}")
        return []

    except requests.exceptions.RequestException as err:
        print(f"Request error occurred: {err}")
        return []


state = "Texas"

breweries_info = fetch_breweries_by_state(state)

if breweries_info:

    for data in breweries_info:
        name = data["name"]
        brewery_type = data["brewery_type"]
        city = data["city"]

        print(f"{name} of {brewery_type} is located in {city}")

else:
    print("No breweries found")
````

### Task 2: Fetch User Data for AtliQ Client Project

**Scenario:** As a quick favor, Loki shared an ad-hoc task with you. He needs help with a user profile component for an Atliq client application, using JSONPlaceholder to simulate fetching user data.

**Objective:** Create a function `fetch_user_data` that uses JSONPlaceholder’s API to retrieve and format user details.

**Steps:**

1. Fetch user data from JSONPlaceholder.
2. Format the data to include 'name', 'email', and 'address'.
3. Return the data as a dictionary.

**Expected Output:**

```
{
    'name': 'Leanne Graham',
    'email': 'Sincere@april.biz',
    'address': 'Kulas Light, Gwenborough'
}



```

````python
# write your code here
def fetch_user_data(user_id):
    url=f'https://jsonplaceholder.typicode.com/users/{user_id}' 
    try:
        response=requests.get(url)
        response.raise_for_status()
        info=response.json()

        if info:
            address=f"{info['address']['street']}, {info['address']['city']}, {info['address']['zipcode']}"
            return {
                'name' : info['name'],
                'email' : info['email'],
                'address' : address
            }
        return "No data Found"
    except requests.exceptions.HTTPError as err: 
        return f"HTTP error occured {err}"
    except requests.exceptions.RequestException as err:
        return f"An error occured {err}"

user_id=3
get_info=fetch_user_data(user_id)
get_info
        

````
