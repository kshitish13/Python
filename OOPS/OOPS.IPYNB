```python
class Car:
    #variables
    brand="Toyota"
    wheels=4

    #functions/methods
    def accelarate(self):
        print("accelarated")
    
#object
car1=Car()
print(car1.brand)
print(car1.wheels)
car1.accelarate()
```

```python
class Factory:
    #constructor
    def __init__(self,material,zips,pockets):
        self.material=material
        self.zips=zips
        self.pockets=pockets
 
#object
reebok=Factory("leather",3,2)
print(reebok.material)
```

Inheritance


```python
class Factory: #parent class / superclass
    a="I am an attribute metioned inside factory."

    def hello(self):
        print("I am an method metioned inside factory.")

class Factorypune(Factory): #child class/ subclass
    pass

obj = Factory()
print(obj.a)

obj2=Factorypune()
obj2.hello()
```

Constructor in Inheritance

```python
class Parent:
    def __init__(self,name):
        self.name=name

class Child(Parent):
    def display(self):
        print(f"My name is {self.name}")

h1=Child("Kshitish")
h1.display()

p1=Parent("Abc")
p1.display()
```

```python
class Parent:
    def __init__(self,name):
        self.name=name

class Child(Parent):
    def display(self):
        print(f"My name is {self.name}")

h1=Child("Kshitish")
h1.display()
```

```python
class Parent:
    def __init__(self,name):
        self.name=name

    def display(self):
        print(f"My name is {self.name}")

class Child(Parent):
    pass

h1=Child("Kshitish")
h1.display()

p1=Parent("Abc")
p1.display()
```

using super keyword for child class constructor

```python
class Father:
    def skills(self):
        print("Coding")


class Mother:
    def skills(self):
        print("Cooking")


class Child(Father, Mother):
    def show(self):
        print("I have multiple skills")


c = Child()

c.show()
c.skills()
```
